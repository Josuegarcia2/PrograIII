```python
iss = lambda s: min(s * 0.03, 30)
afp = lambda s: s * 0.0725
isr = lambda b: 0 if b <= 550 else (b - 550) * 0.10 + 17.67 if b <= 895.24 else (b - 895.24) * 0.20 + 60.00 if b <= 2038.10 else (b - 2038.10) * 0.30 + 288.57
neto = lambda s: s - iss(s) - afp(s) - isr(s - iss(s) - afp(s))
```

```python
# Determinar si un número es par, impar o primo
continuar = "s"

while continuar.lower() == "s":
    try:
        numero = int(input("Ingresa un número: "))
    except KeyboardInterrupt:
        print("\nPrograma interrumpido por el usuario.")
        break
    except ValueError:
        print("Debes ingresar un número entero. Intenta de nuevo.")
        continue


    if numero % 2 == 0:
        print("El número", numero, "es par")
    else:
        print("El número", numero, "es impar")


    primo = True
    if numero <= 1:
        primo = False
    else:

        for divisor in range(2, numero // 2 + 1):
            if numero % divisor == 0:
                primo = False
                break

    if primo:
        print("Y además es primo")
    else:
        print("Y no es primo")

    try:
        continuar = input("¿Continuar? (s para sí, otra tecla para no): ")
    except KeyboardInterrupt:
        print("\nPrograma interrumpido por el usuario.")
        break

print("\nFin del programa")
```

**Ejercicio 1: Cálculo de deducciones de sueldo (ISSS, AFP, ISR)**

```python
# Deducciones de sueldo con lambdas y función para ISR
def calcular_isr(sueldo_gravado):
    """Calcula ISR según tabla progresiva (El Salvador)."""
    if sueldo_gravado <= 472.00:
        return 0
    elif sueldo_gravado <= 895.24:
        return (sueldo_gravado - 472.00) * 0.10
    elif sueldo_gravado <= 2038.10:
        return (sueldo_gravado - 895.24) * 0.20 + 42.32
    elif sueldo_gravado <= 3819.79:
        return (sueldo_gravado - 2038.10) * 0.30 + 270.89
    else:
        return (sueldo_gravado - 3819.79) * 0.30 + 804.68

# Lambdas
iss = lambda s: s * 0.03
afp = lambda s: s * 0.0625

# Cálculo del neto
def sueldo_neto(sueldo_bruto):
    deduccion_iss = iss(sueldo_bruto)
    deduccion_afp = afp(sueldo_bruto)
    base_isr = sueldo_bruto - deduccion_iss - deduccion_afp
    deduccion_isr = calcular_isr(base_isr)
    return sueldo_bruto - deduccion_iss - deduccion_afp - deduccion_isr

# Programa principal
try:
    sueldo = float(input("Ingrese su sueldo bruto mensual: $"))
except ValueError:
    print("Debe ingresar un número.")
else:
    print(f"ISSS: ${iss(sueldo):.2f}")
    print(f"AFP: ${afp(sueldo):.2f}")
    print(f"ISR: ${calcular_isr(sueldo - iss(sueldo) - afp(sueldo)):.2f}")
    print(f"Sueldo neto: ${sueldo_neto(sueldo):.2f}")
```

**Ejercicio 2 – Pago de agua potable**

```python
# Tarifa de agua
def calcular_pago(metros):
    if metros <= 18:
        return 6.0
    elif metros <= 28:
        return 6.0 + (metros - 18) * 0.45
    else:
        return 6.0 + (28 - 18) * 0.45 + (metros - 28) * 0.65

# Programa
try:
    consumo = float(input("Metros cúbicos consumidos: "))
except ValueError:
    print("Ingrese un valor numérico.")
else:
    pago = calcular_pago(consumo)
    print(f"Consumo: {consumo} mts, Total a pagar: ${pago:.2f}")
```

**Ejercicio 4 – Conversiones (monedas, longitud, masa, almacenamiento, tiempo, volumen, área)**

```python
# CONVERSOR DE MONEDAS

monedas = {
    1: ("Dólar", 1.0),
    2: ("Euro", 0.92),
    3: ("Libra esterlina", 0.78),
    4: ("Yen japonés", 147.50),
    5: ("Peso mexicano", 17.20),
    6: ("Colón costarricense", 525.00),
    7: ("Sol peruano", 3.75),
    8: ("Real brasileño", 5.10),
    9: ("Peso argentino", 870.00),
    10: ("Peso chileno", 920.00),
    11: ("Bolívar venezolano", 36.50),
    12: ("Peso colombiano", 4000.00)
}

while True:
    print("\n--- CONVERSOR DE MONEDAS ---")
    print("Monedas disponibles:")
    for clave, (nombre, _) in monedas.items():
        print(f"  {clave}. {nombre}")

    try:
        origen = int(input("\nSeleccione la moneda de origen (número): "))
        destino = int(input("Seleccione la moneda de destino (número): "))
    except ValueError:
        print("Entrada inválida. Debe ser un número.")
        continue

    if origen not in monedas or destino not in monedas:
        print("Opción no válida. Intente de nuevo.")
        continue

    if origen == destino:
        print("La moneda de origen y destino son iguales. No hay conversión.")
        continue

    try:
        cantidad = float(input(f"Cantidad en {monedas[origen][0]}: "))
    except ValueError:
        print("Cantidad inválida.")
        continue

    valor_en_dolar = cantidad / monedas[origen][1]
    resultado = valor_en_dolar * monedas[destino][1]

    print(f"{cantidad} {monedas[origen][0]} = {resultado:.4f} {monedas[destino][0]}")

    continuar = input("¿Otra conversión? (s/n): ").lower()
    if continuar != 's':
        print("¡Hasta luego!")
        break
```

**Ejercicio 5: Realizar el sistema de gestión del tiempo y la demanda en las unidades de salud.**

```python
from functools import reduce

especialidades = {
    1: "Medicina General",
    2: "Pediatría",
    3: "Ginecología",
    4: "Cardiología"
}

medicos = {
    "Medicina General": ["Dr. Pérez", "Dra. Gómez"],
    "Pediatría": ["Dra. Martínez", "Dr. Rodríguez"],
    "Ginecología": ["Dra. López", "Dr. Sánchez"],
    "Cardiología": ["Dr. Ramírez", "Dra. Torres"]
}

horarios = ["08:00", "09:00", "10:00", "11:00", "14:00", "15:00", "16:00"]

citas = []

def menu():
    print("\n" + "=" * 40)
    print("   UNIDAD DE SALUD")
    print("=" * 40)
    print("1. Agendar cita")
    print("2. Ver citas")
    print("3. Buscar paciente")
    print("4. Estadísticas")
    print("5. Reporte")
    print("6. Salir")
    print("=" * 40)

def agendar():
    print("\n--- AGENDAR CITA ---")

    try:
        nombre = input("Paciente: ")
        telefono = input("Teléfono: ")

        print("\nEspecialidades:")
        for codigo, esp in especialidades.items():
            print(f"  {codigo}. {esp}")

        esp_cod = int(input("Elija especialidad: "))
        especialidad = especialidades[esp_cod]

        print(f"\nMédicos de {especialidad}:")
        for i, med in enumerate(medicos[especialidad], 1):
            print(f"  {i}. {med}")

        med_idx = int(input("Elija médico: ")) - 1
        medico = medicos[especialidad][med_idx]

        fecha = input("Fecha (dd/mm/aaaa): ")

        print("\nHorarios disponibles:")
        horarios_libres = list(filter(lambda h: h not in [c["hora"] for c in citas if c["fecha"] == fecha], horarios))
        for i, hora in enumerate(horarios_libres, 1):
            print(f"  {i}. {hora}")

        hora_idx = int(input("Elija hora: ")) - 1
        hora = horarios_libres[hora_idx]

        cita = {
            "paciente": nombre,
            "telefono": telefono,
            "especialidad": especialidad,
            "medico": medico,
            "fecha": fecha,
            "hora": hora,
            "estado": "Activa"
        }

        citas.append(cita)
        print(f"\n✅ Cita agendada: {nombre} - {fecha} {hora}")

    except ValueError:
        print("⚠️ Error: Ingrese números válidos.")
    except:
        print("⚠️ Error al agendar.")

def ver_citas():
    if not citas:
        print("\n⚠️ No hay citas.")
        return

    print("\n" + "=" * 50)
    print("   LISTA DE CITAS")
    print("=" * 50)

    for i, c in enumerate(citas, 1):
        print(f"{i}. {c['paciente']} | {c['fecha']} {c['hora']} | {c['especialidad']} | {c['medico']} | {c['estado']}")

def buscar():
    if not citas:
        print("\n⚠️ No hay citas.")
        return

    nombre = input("Nombre del paciente: ").lower()

    encontrados = list(filter(lambda c: nombre in c["paciente"].lower(), citas))

    if encontrados:
        print(f"\n📋 Citas de {nombre}:")
        for c in encontrados:
            print(f"  • {c['paciente']} - {c['fecha']} {c['hora']} - {c['especialidad']}")
    else:
        print(f"\n⚠️ No se encontraron citas.")

def estadisticas():
    if not citas:
        print("\n⚠️ No hay citas.")
        return

    print("\n" + "=" * 40)
    print("   ESTADÍSTICAS")
    print("=" * 40)

    total = len(citas)
    activas = len(list(filter(lambda c: c["estado"] == "Activa", citas)))

    print(f"\nTotal citas: {total}")
    print(f"Citas activas: {activas}")
    print(f"Citas canceladas: {total - activas}")

    print("\nPor especialidad:")
    for esp in especialidades.values():
        conteo = len(list(filter(lambda c: c["especialidad"] == esp, citas)))
        if conteo > 0:
            print(f"  {esp}: {conteo}")

    print("\nPor médico:")
    todos_medicos = list(set(map(lambda c: c["medico"], citas)))
    for med in todos_medicos:
        conteo = len(list(filter(lambda c: c["medico"] == med, citas)))
        print(f"  {med}: {conteo}")

def reporte():
    if not citas:
        print("\n⚠️ No hay citas.")
        return

    print("\n" + "=" * 40)
    print("   REPORTE DE OCUPACIÓN")
    print("=" * 40)

    for esp in especialidades.values():
        total_esp = len(list(filter(lambda c: c["especialidad"] == esp, citas)))
        activas_esp = len(list(filter(lambda c: c["especialidad"] == esp and c["estado"] == "Activa", citas)))

        if total_esp > 0:
            print(f"\n{esp}:")
            print(f"  Total: {total_esp}")
            print(f"  Activas: {activas_esp}")
            print(f"  Ocupación: {activas_esp}/{total_esp}")

def cargar_ejemplos():
    ejemplos = [
        {"paciente": "Ana García", "telefono": "7890-1234", "especialidad": "Medicina General",
         "medico": "Dr. Pérez", "fecha": "20/08/2026", "hora": "09:00", "estado": "Activa"},
        {"paciente": "Carlos Ramírez", "telefono": "7890-5678", "especialidad": "Pediatría",
         "medico": "Dra. Martínez", "fecha": "21/08/2026", "hora": "10:00", "estado": "Activa"},
        {"paciente": "María López", "telefono": "7890-9012", "especialidad": "Ginecología",
         "medico": "Dra. López", "fecha": "20/08/2026", "hora": "14:00", "estado": "Activa"},
        {"paciente": "Juan Pérez", "telefono": "7890-3456", "especialidad": "Cardiología",
         "medico": "Dr. Ramírez", "fecha": "22/08/2026", "hora": "11:00", "estado": "Cancelada"}
    ]

    for e in ejemplos:
        citas.append(e)

    print(f"✅ {len(ejemplos)} citas de ejemplo cargadas.")

if __name__ == "__main__":
    cargar_ejemplos()

    while True:
        menu()

        try:
            opcion = input("Opción: ")
        except KeyboardInterrupt:
            print("\n\n👋 ¡Hasta luego!")
            break

        if opcion == "1":
            agendar()
        elif opcion == "2":
            ver_citas()
        elif opcion == "3":
            buscar()
        elif opcion == "4":
            estadisticas()
        elif opcion == "5":
            reporte()
        elif opcion == "6":
            print("\n👋 ¡Hasta luego!")
            break
        else:
            print("⚠️ Opción no válida.")
```
