```python
iss = lambda s: min(s * 0.03, 30)
afp = lambda s: s * 0.0725
isr = lambda b: 0 if b <= 550 else (b - 550) * 0.10 + 17.67 if b <= 895.24 else (b - 895.24) * 0.20 + 60.00 if b <= 2038.10 else (b - 2038.10) * 0.30 + 288.57
neto = lambda s: s - iss(s) - afp(s) - isr(s - iss(s) - afp(s))
```

```python
# Determinar si un número es par, impar o primo
continuar = "s"

while continuar.lower() == "s":
    try:
        numero = int(input("Ingresa un número: "))
    except KeyboardInterrupt:
        print("\nPrograma interrumpido por el usuario.")
        break
    except ValueError:
        print("Debes ingresar un número entero. Intenta de nuevo.")
        continue


    if numero % 2 == 0:
        print("El número", numero, "es par")
    else:
        print("El número", numero, "es impar")


    primo = True
    if numero <= 1:
        primo = False
    else:

        for divisor in range(2, numero // 2 + 1):
            if numero % divisor == 0:
                primo = False
                break

    if primo:
        print("Y además es primo")
    else:
        print("Y no es primo")

    try:
        continuar = input("¿Continuar? (s para sí, otra tecla para no): ")
    except KeyboardInterrupt:
        print("\nPrograma interrumpido por el usuario.")
        break

print("\nFin del programa")
```

**Ejercicio 1: Cálculo de deducciones de sueldo (ISSS, AFP, ISR)**

```python
# Deducciones de sueldo con lambdas y función para ISR
def calcular_isr(sueldo_gravado):
    """Calcula ISR según tabla progresiva (El Salvador)."""
    if sueldo_gravado <= 472.00:
        return 0
    elif sueldo_gravado <= 895.24:
        return (sueldo_gravado - 472.00) * 0.10
    elif sueldo_gravado <= 2038.10:
        return (sueldo_gravado - 895.24) * 0.20 + 42.32
    elif sueldo_gravado <= 3819.79:
        return (sueldo_gravado - 2038.10) * 0.30 + 270.89
    else:
        return (sueldo_gravado - 3819.79) * 0.30 + 804.68

# Lambdas
iss = lambda s: s * 0.03
afp = lambda s: s * 0.0625

# Cálculo del neto
def sueldo_neto(sueldo_bruto):
    deduccion_iss = iss(sueldo_bruto)
    deduccion_afp = afp(sueldo_bruto)
    base_isr = sueldo_bruto - deduccion_iss - deduccion_afp
    deduccion_isr = calcular_isr(base_isr)
    return sueldo_bruto - deduccion_iss - deduccion_afp - deduccion_isr

# Programa principal
try:
    sueldo = float(input("Ingrese su sueldo bruto mensual: $"))
except ValueError:
    print("Debe ingresar un número.")
else:
    print(f"ISSS: ${iss(sueldo):.2f}")
    print(f"AFP: ${afp(sueldo):.2f}")
    print(f"ISR: ${calcular_isr(sueldo - iss(sueldo) - afp(sueldo)):.2f}")
    print(f"Sueldo neto: ${sueldo_neto(sueldo):.2f}")
```

**Ejercicio 2 – Pago de agua potable**

```python
# Tarifa de agua
def calcular_pago(metros):
    if metros <= 18:
        return 6.0
    elif metros <= 28:
        return 6.0 + (metros - 18) * 0.45
    else:
        return 6.0 + (28 - 18) * 0.45 + (metros - 28) * 0.65

# Programa
try:
    consumo = float(input("Metros cúbicos consumidos: "))
except ValueError:
    print("Ingrese un valor numérico.")
else:
    pago = calcular_pago(consumo)
    print(f"Consumo: {consumo} mts, Total a pagar: ${pago:.2f}")
```

**Ejercicio 4 – Conversiones (monedas, longitud, masa, almacenamiento, tiempo, volumen, área)**

```python
# CONVERSOR DE MONEDAS

monedas = {
    1: ("Dólar", 1.0),
    2: ("Euro", 0.92),
    3: ("Libra esterlina", 0.78),
    4: ("Yen japonés", 147.50),
    5: ("Peso mexicano", 17.20),
    6: ("Colón costarricense", 525.00),
    7: ("Sol peruano", 3.75),
    8: ("Real brasileño", 5.10),
    9: ("Peso argentino", 870.00),
    10: ("Peso chileno", 920.00),
    11: ("Bolívar venezolano", 36.50),
    12: ("Peso colombiano", 4000.00)
}

while True:
    print("\n--- CONVERSOR DE MONEDAS ---")
    print("Monedas disponibles:")
    for clave, (nombre, _) in monedas.items():
        print(f"  {clave}. {nombre}")

    try:
        origen = int(input("\nSeleccione la moneda de origen (número): "))
        destino = int(input("Seleccione la moneda de destino (número): "))
    except ValueError:
        print("Entrada inválida. Debe ser un número.")
        continue

    if origen not in monedas or destino not in monedas:
        print("Opción no válida. Intente de nuevo.")
        continue

    if origen == destino:
        print("La moneda de origen y destino son iguales. No hay conversión.")
        continue

    try:
        cantidad = float(input(f"Cantidad en {monedas[origen][0]}: "))
    except ValueError:
        print("Cantidad inválida.")
        continue

    valor_en_dolar = cantidad / monedas[origen][1]
    resultado = valor_en_dolar * monedas[destino][1]

    print(f"{cantidad} {monedas[origen][0]} = {resultado:.4f} {monedas[destino][0]}")

    continuar = input("¿Otra conversión? (s/n): ").lower()
    if continuar != 's':
        print("¡Hasta luego!")
        break
```

```python
# SISTEMA DE PRIORIZACIÓN DE PACIENTES

prioridades = {
    1: "🔴 PRIORIDAD I - ROJO",
    2: "🟡 PRIORIDAD II - AMARILLO",
    3: "🟢 PRIORIDAD III - VERDE"
}

tiempos_atencion = {
    "🔴 PRIORIDAD I - ROJO": "ATENCIÓN INMEDIATA",
    "🟡 PRIORIDAD II - AMARILLO": "ATENCIÓN EN MENOS DE 2 HORAS",
    "🟢 PRIORIDAD III - VERDE": "ATENCIÓN EN 2 A 4 HORAS"
}

colores = {
    "🔴 PRIORIDAD I - ROJO": "ROJO",
    "🟡 PRIORIDAD II - AMARILLO": "AMARILLO",
    "🟢 PRIORIDAD III - VERDE": "VERDE"
}

pacientes = []

def menu_principal():
    print("\n" + "=" * 50)
    print("   SISTEMA DE PRIORIZACIÓN DE PACIENTES")
    print("=" * 50)
    print("1. Registrar nuevo paciente")
    print("2. Ver lista de pacientes")
    print("3. Buscar paciente")
    print("4. Estadísticas de prioridades")
    print("5. Reporte completo")
    print("6. Salir")
    print("=" * 50)

def asignar_prioridad(edad, sintomas):
    inconsciente = sintomas.get("inconsciente", False)
    respirar = sintomas.get("respirar", False)
    pecho = sintomas.get("pecho", False)
    sangrado = sintomas.get("sangrado", False)
    obstetrica = sintomas.get("obstetrica", False)
    convulsiones = sintomas.get("convulsiones", False)
    fiebre_alta = sintomas.get("fiebre_alta", False)
    deshidratacion = sintomas.get("deshidratacion", False)
    embarazada = sintomas.get("embarazada", False)
    enfermedad_cronica = sintomas.get("enfermedad_cronica", False)
    control_preventivo = sintomas.get("control_preventivo", False)
    enfermedad_leve = sintomas.get("enfermedad_leve", False)

    if (inconsciente or respirar or pecho or sangrado or
        obstetrica or convulsiones or
        (fiebre_alta and edad < 10) or
        (edad >= 60 and enfermedad_cronica and fiebre_alta) or
        (embarazada and (sangrado or fiebre_alta or convulsiones)) or
        edad >= 85):

        prioridad = "🔴 PRIORIDAD I - ROJO"
        motivo = "Paciente presenta síntomas graves o emergencia obstétrica que requiere atención inmediata"

    elif (edad >= 60 or embarazada or enfermedad_cronica or
          edad < 1 or (edad < 10 and fiebre_alta) or deshidratacion):

        prioridad = "🟡 PRIORIDAD II - AMARILLO"
        motivo = "Paciente en grupo de riesgo por edad, embarazo o condición médica"

    else:
        prioridad = "🟢 PRIORIDAD III - VERDE"
        motivo = "Paciente con síntomas leves o atención preventiva"

    return prioridad, motivo

def registrar_paciente():
    print("\n" + "-" * 40)
    print("   REGISTRO DE NUEVO PACIENTE")
    print("-" * 40)

    try:
        nombre = input("Nombre completo: ")

        while True:
            try:
                edad = int(input("Edad (años): "))
                if edad < 0 or edad > 130:
                    print(" Edad no válida. Debe ser entre 0 y 130 años.")
                    continue
                break
            except ValueError:
                print(" Por favor, ingrese un número válido.")

        telefono = input("Teléfono: ")
        direccion = input("Dirección: ")

        print("\n--- SÍNTOMAS Y CONDICIONES ---")
        print("Responda 's' para Sí o 'n' para No")
        print("-" * 40)

        sintomas = {
            "inconsciente": input("¿Está inconsciente o no responde? (s/n): ").lower() == 's',
            "respirar": input("¿Tiene dificultad para respirar? (s/n): ").lower() == 's',
            "pecho": input("¿Tiene dolor en el pecho o palpitaciones? (s/n): ").lower() == 's',
            "sangrado": input("¿Tiene sangrado abundante? (s/n): ").lower() == 's',
            "obstetrica": input("¿Es una emergencia obstétrica (parto, sangrado)? (s/n): ").lower() == 's',
            "convulsiones": input("¿Tiene convulsiones o ataques? (s/n): ").lower() == 's',
            "fiebre_alta": input("¿Tiene fiebre muy alta (mayor a 39°C)? (s/n): ").lower() == 's',
            "deshidratacion": input("¿Tiene deshidratación severa? (s/n): ").lower() == 's',
            "embarazada": input("¿Está embarazada? (s/n): ").lower() == 's',
            "enfermedad_cronica": input("¿Tiene enfermedad crónica (diabetes, hipertensión)? (s/n): ").lower() == 's',
            "control_preventivo": input("¿Solo necesita control o chequeo preventivo? (s/n): ").lower() == 's',
            "enfermedad_leve": input("¿Tiene enfermedad leve / resfriado / malestar? (s/n): ").lower() == 's'
        }

        prioridad, motivo = asignar_prioridad(edad, sintomas)

        paciente = {
            "nombre": nombre,
            "edad": edad,
            "telefono": telefono,
            "direccion": direccion,
            "sintomas": sintomas,
            "prioridad": prioridad,
            "motivo": motivo,
            "tiempo_atencion": tiempos_atencion[prioridad],
            "color": colores[prioridad]
        }

        pacientes.append(paciente)

        print("\n" + "=" * 50)
        print("   PRIORIDAD ASIGNADA")
        print("=" * 50)
        print(f"\n{paciente['prioridad']}")
        print(f"⏰ {paciente['tiempo_atencion']}")
        print(f"\n📋 PACIENTE: {nombre}")
        print(f"📅 EDAD: {edad} años")
        print(f"📞 TELÉFONO: {telefono}")
        print(f"📍 DIRECCIÓN: {direccion}")
        print(f"\n💬 MOTIVO: {motivo}")
        print("\n" + "=" * 50)

    except KeyboardInterrupt:
        print("\n\n Operación cancelada.")
    except Exception as e:
        print(f"\n Error al registrar: {e}")

def ver_pacientes():
    if not pacientes:
        print("\n📋 No hay pacientes registrados.")
        return

    print("\n" + "=" * 60)
    print("   LISTA DE PACIENTES REGISTRADOS")
    print("=" * 60)

    orden_prioridad = {
        "🔴 PRIORIDAD I - ROJO": 1,
        "🟡 PRIORIDAD II - AMARILLO": 2,
        "🟢 PRIORIDAD III - VERDE": 3
    }

    pacientes_ordenados = sorted(pacientes, key=lambda p: (orden_prioridad[p['prioridad']], -p['edad']))

    for i, p in enumerate(pacientes_ordenados, 1):
        print(f"{i}. {p['nombre']} | {p['edad']} años | {p['prioridad']} | {p['tiempo_atencion']}")

def buscar_paciente():
    if not pacientes:
        print("\n📋 No hay pacientes registrados.")
        return

    nombre_buscar = input("\nNombre del paciente a buscar: ").lower()

    encontrados = list(filter(lambda p: nombre_buscar in p['nombre'].lower(), pacientes))

    if encontrados:
        print(f"\n📋 Pacientes encontrados ({len(encontrados)}):")
        print("-" * 50)

        for p in encontrados:
            print(f"\n👤 {p['nombre']}")
            print(f"📅 {p['edad']} años")
            print(f"📞 {p['telefono']}")
            print(f"📍 {p['direccion']}")
            print(f"🔴 {p['prioridad']}")
            print(f"⏰ {p['tiempo_atencion']}")
            print(f"💬 {p['motivo']}")
            print("-" * 30)
    else:
        print(f"\n❌ No se encontraron pacientes con el nombre '{nombre_buscar}'")

def estadisticas():
    if not pacientes:
        print("\n📋 No hay pacientes registrados.")
        return

    print("\n" + "=" * 50)
    print("   ESTADÍSTICAS DE PRIORIDADES")
    print("=" * 50)

    total = len(pacientes)
    print(f"\n📊 TOTAL DE PACIENTES: {total}")

    rojos = len(list(filter(lambda p: p['prioridad'] == "🔴 PRIORIDAD I - ROJO", pacientes)))
    amarillos = len(list(filter(lambda p: p['prioridad'] == "🟡 PRIORIDAD II - AMARILLO", pacientes)))
    verdes = len(list(filter(lambda p: p['prioridad'] == "🟢 PRIORIDAD III - VERDE", pacientes)))

    print(f"\n🔴 PRIORIDAD I (ROJO): {rojos} pacientes")
    print(f"🟡 PRIORIDAD II (AMARILLO): {amarillos} pacientes")
    print(f"🟢 PRIORIDAD III (VERDE): {verdes} pacientes")

    if total > 0:
        print(f"\n📈 PORCENTAJES:")
        print(f"   Rojo: {(rojos/total)*100:.1f}%")
        print(f"   Amarillo: {(amarillos/total)*100:.1f}%")
        print(f"   Verde: {(verdes/total)*100:.1f}%")

    edades = list(map(lambda p: p['edad'], pacientes))
    if edades:
        print(f"\n📊 ESTADÍSTICAS DE EDAD:")
        print(f"   Edad promedio: {sum(edades)/len(edades):.1f} años")
        print(f"   Edad mínima: {min(edades)} años")
        print(f"   Edad máxima: {max(edades)} años")

def reporte_completo():
    if not pacientes:
        print("\n📋 No hay pacientes registrados.")
        return

    print("\n" + "=" * 60)
    print("   REPORTE COMPLETO DE PACIENTES")
    print("=" * 60)

    orden_prioridad = {
        "🔴 PRIORIDAD I - ROJO": 1,
        "🟡 PRIORIDAD II - AMARILLO": 2,
        "🟢 PRIORIDAD III - VERDE": 3
    }

    pacientes_ordenados = sorted(pacientes, key=lambda p: (orden_prioridad[p['prioridad']], -p['edad']))

    for i, p in enumerate(pacientes_ordenados, 1):
        print(f"\n{i}. {p['prioridad']}")
        print(f"   👤 {p['nombre']} - {p['edad']} años")
        print(f"   📞 {p['telefono']} | 📍 {p['direccion']}")
        print(f"   ⏰ {p['tiempo_atencion']}")
        print(f"   💬 {p['motivo']}")

def cargar_pacientes_ejemplo():
    ejemplos = [
        {
            "nombre": "Josue Elias Garcia Lopez",
            "edad": 22,
            "telefono": "70009403",
            "direccion": "Puerto El Triunfo",
            "sintomas": {"inconsciente": True, "sangrado": True, "enfermedad_cronica": True, "control_preventivo": True},
            "prioridad": "🔴 PRIORIDAD I - ROJO",
            "motivo": "Paciente presenta síntomas graves o emergencia obstétrica que requiere atención inmediata",
            "tiempo_atencion": "ATENCIÓN INMEDIATA",
            "color": "ROJO"
        },
        {
            "nombre": "Ana García",
            "edad": 28,
            "telefono": "7890-9012",
            "direccion": "Colonia San Miguel #789",
            "sintomas": {"embarazada": True, "sangrado": True},
            "prioridad": "🔴 PRIORIDAD I - ROJO",
            "motivo": "Paciente presenta síntomas graves o emergencia obstétrica que requiere atención inmediata",
            "tiempo_atencion": "ATENCIÓN INMEDIATA",
            "color": "ROJO"
        },
        {
            "nombre": "María Pérez",
            "edad": 70,
            "telefono": "7890-1234",
            "direccion": "Calle Los Ángeles #123",
            "sintomas": {"enfermedad_cronica": True, "fiebre_alta": True},
            "prioridad": "🔴 PRIORIDAD I - ROJO",
            "motivo": "Paciente presenta síntomas graves o emergencia obstétrica que requiere atención inmediata",
            "tiempo_atencion": "ATENCIÓN INMEDIATA",
            "color": "ROJO"
        },
        {
            "nombre": "Carlos Ramírez",
            "edad": 45,
            "telefono": "7890-5678",
            "direccion": "Av. Las Palmas #456",
            "sintomas": {"pecho": True},
            "prioridad": "🔴 PRIORIDAD I - ROJO",
            "motivo": "Paciente presenta síntomas graves o emergencia obstétrica que requiere atención inmediata",
            "tiempo_atencion": "ATENCIÓN INMEDIATA",
            "color": "ROJO"
        },
        {
            "nombre": "Laura Martínez",
            "edad": 30,
            "telefono": "7890-3456",
            "direccion": "Pasaje La Paz #101",
            "sintomas": {"embarazada": True, "enfermedad_cronica": True},
            "prioridad": "🟡 PRIORIDAD II - AMARILLO",
            "motivo": "Paciente en grupo de riesgo por edad, embarazo o condición médica",
            "tiempo_atencion": "ATENCIÓN EN MENOS DE 2 HORAS",
            "color": "AMARILLO"
        },
        {
            "nombre": "Luis Martínez",
            "edad": 25,
            "telefono": "7890-3456",
            "direccion": "Pasaje La Paz #101",
            "sintomas": {"enfermedad_leve": True},
            "prioridad": "🟢 PRIORIDAD III - VERDE",
            "motivo": "Paciente con síntomas leves o atención preventiva",
            "tiempo_atencion": "ATENCIÓN EN 2 A 4 HORAS",
            "color": "VERDE"
        }
    ]

    for e in ejemplos:
        pacientes.append(e)

    print(f"✅ {len(ejemplos)} pacientes de ejemplo cargados.")

if __name__ == "__main__":
    cargar_pacientes_ejemplo()

    while True:
        try:
            menu_principal()
            opcion = input("Seleccione una opción: ")

            if opcion == "1":
                registrar_paciente()
            elif opcion == "2":
                ver_pacientes()
            elif opcion == "3":
                buscar_paciente()
            elif opcion == "4":
                estadisticas()
            elif opcion == "5":
                reporte_completo()
            elif opcion == "6":
                print("\n👋 ¡Hasta luego!")
                break
            else:
                print("\n⚠️ Opción no válida. Intente de nuevo.")

        except KeyboardInterrupt:
            print("\n\n👋 ¡Hasta luego!")
            break
        except Exception as e:
            print(f"\n⚠️ Error: {e}")
```

**Hacer un programa en python que lea un serie  de numeros entroducida por el usuario y que calcule la media aritmetica**

**TAREA**


1.   Econtrar la media armonica
2.   Encontrar la frecuencia
3.   Escriba un programa en python que determine si es una palabra introducida por el usuario es palindrome. Ej: ana, oso, oro, reconocer, bob, somos, salas, ala, etc. Haciendo uso de clases, objetos, funciones, etc.



```python
from functools import reduce
from collections import Counter

class estadistica:

    def media(self, serie):
        suma = reduce(lambda acum, num: acum + float(num), serie, 0)
        return suma / len(serie)

    def varianza(self, serie):
        media = self.media(serie)
        suma = reduce(lambda acum, num: acum + (float(num) - media) ** 2, serie, 0)
        return suma / len(serie)

    def desviacion(self, serie):
        return self.varianza(serie) ** 0.5

    def media_armonica(self, serie):
        n = len(serie)
        suma_reciprocos = reduce(lambda acum, num: acum + (1 / float(num)), serie, 0)
        return n / suma_reciprocos

    def frecuencia(self, serie):
        numeros = [float(num) for num in serie]
        contador = Counter(numeros)
        return dict(sorted(contador.items()))


class Palindromo:

    def limpiar(self, palabra):
        return palabra.lower().replace(" ", "")

    def es_palindromo(self, palabra):
        limpia = self.limpiar(palabra)
        return limpia == limpia[::-1]

    def verificar(self, palabra):
        if self.es_palindromo(palabra):
            print(f'"{palabra}" SI es un palindromo.')
        else:
            print(f'"{palabra}" NO es un palindromo.')
```

```python
# ESTADISTICA
objEstadistica = estadistica()
serie = input("Serie: ").split(",")

print("\n Resultados Estadistica: ")
print("Media: ",          objEstadistica.media(serie))
print("Varianza: ",       objEstadistica.varianza(serie))
print("Desviacion: ",     objEstadistica.desviacion(serie))
print("Media Armonica: ", objEstadistica.media_armonica(serie))
print("Frecuencia: ",     objEstadistica.frecuencia(serie))

# PALINDROMO
print("\n Verificador de Palindromo: ")
objPalindromo = Palindromo()
palabra = input("Ingresa una palabra: ")
objPalindromo.verificar(palabra)
```
