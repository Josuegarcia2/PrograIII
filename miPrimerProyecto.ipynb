Calcular el sueldo de un empleado. hrs * valor de horas * 30 dias las deducciones: ISSS, AFP, ISR (Ver tabla del MH)

```python
iss = lambda s: min(s * 0.03, 30)
afp = lambda s: s * 0.0725
isr = lambda b: 0 if b <= 550 else (b - 550) * 0.10 + 17.67 if b <= 895.24 else (b - 895.24) * 0.20 + 60.00 if b <= 2038.10 else (b - 2038.10) * 0.30 + 288.57
neto = lambda s: s - iss(s) - afp(s) - isr(s - iss(s) - afp(s))
```

```python
s = float(input("Sueldo bruto mensual: $"))
```

```python
print(f"ISSS: ${iss(s):.2f}")
print(f"AFP: ${afp(s):.2f}")
print(f"ISR: ${isr(s - iss(s) - afp(s)):.2f}")
print(f"Sueldo neto: ${neto(s):.2f}")
```

```python
#FILTER
#dada una serie numero dame solo lso que son pares
serie = [2,5,6,8,9,10,12]

pares = filter(lambda num: num%2==0, serie)
impares = filter(lambda num: num%2!=0, serie)
print("Original: ", serie)
print("Pares: ", list(pares))
print("Impares: ", list(impares))
```

```python
# Dada una serie de notas, clasificar los aprobados y reprobados.

notas = [4,7,10,9,3,6,5,7,8,4,6]

Aprobacion  = 6

aprobados = filter(lambda nota: nota >= Aprobacion, notas)
reprobados = filter(lambda nota: nota < Aprobacion, notas)


print("Notas originales: ", notas)
print("Aprobados: ", list(aprobados))
print("Reprobados: ", list(reprobados))
```

```python
#reduce
#Cuanto es el acumulado de los numeros pares de una serie
from functools import reduce

numeros = [1,2,8,4,9,6]
pares = filter(lambda num: num%2==0, numeros)

print("Numeros:", numeros)
print("pares:", list(pares))

sumaPares = reduce(lambda acum, num: num+acum, pares, 0)
print("suma de pares:", sumaPares)
```

**Ejercicio** encontrar la nota final de cada materia

```python
def promedio_computo(notas_computo):
    return (notas_computo["lab1"] + notas_computo["lab2"] + notas_computo["parcial1"]) / 3

def nota_final_materia(materia):
    computos = materia["notas"].values()
    promedios = map(promedio_computo, computos)
    return sum(promedios) / len(materia["notas"])

# Bucle for externo para imprimir
print("Notas finales de", persona["nombre"])
for materia in persona["materias"]:
    nota = nota_final_materia(materia)
    print(f"Materia: {materia['nombre']} (Código {materia['codigo']}) - Nota final: {nota:.2f}")
```

**ESTRUCTURAS DE CONTROL**

```python

seguir = True

while seguir:
    try:
        edad = int(input("Edad: "))
    except KeyboardInterrupt:
        print("\nPrograma interrumpido por el usuario.")
        break
    except ValueError:
        print("Por favor, ingresa un número válido.")
        continue

    if edad < 0:
        print("Edad incorrecta")
        seguir = False
    elif edad <= 2:
        print("Eres un bebé")
    elif edad <= 11:
        print("Eres un niño")
    elif edad <= 17:
        print("Eres un adolescente")
    elif edad <= 64:
        print("Eres un adulto")
    elif edad <= 85:
        print("Eres un adulto mayor")
    elif edad <= 130:
        print("Larga vida")
    else:
        print("Edad incorrecta")
        seguir = False
```

```python

continuar = "s"

while continuar.lower() == "s":
    try:
        ntabla = int(input("Tabla: "))
    except KeyboardInterrupt:
        print("\nPrograma detenido por el usuario.")
        break
    except ValueError:
        print("Debes ingresar un número entero. Intenta de nuevo.")
        continue

    for i in range(1, 11):
        print(ntabla, "x", i, "=", i * ntabla)

    continuar = input("¿Continuar? (s para sí, otra tecla para no): ")

print("Fin del programa.")
```

haciendo uso de las estructuras de control, determinar si un numero introducido
por el usuario es par, impar, o primo.

```python
# Determinar si un número es par, impar o primo
continuar = "s"

while continuar.lower() == "s":
    try:
        numero = int(input("Ingresa un número: "))
    except KeyboardInterrupt:
        print("\nPrograma interrumpido por el usuario.")
        break
    except ValueError:
        print("Debes ingresar un número entero. Intenta de nuevo.")
        continue


    if numero % 2 == 0:
        print("El número", numero, "es par")
    else:
        print("El número", numero, "es impar")


    primo = True
    if numero <= 1:
        primo = False
    else:

        for divisor in range(2, numero // 2 + 1):
            if numero % divisor == 0:
                primo = False
                break

    if primo:
        print("Y además es primo")
    else:
        print("Y no es primo")

    try:
        continuar = input("¿Continuar? (s para sí, otra tecla para no): ")
    except KeyboardInterrupt:
        print("\nPrograma interrumpido por el usuario.")
        break

print("\nFin del programa")
```
